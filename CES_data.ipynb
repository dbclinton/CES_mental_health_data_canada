```python
# Regression code (from Grok):
import statsmodels.formula.api as smf

# 1. Combine the two groups
combined = pd.concat([far_left, far_right], ignore_index=True)

# 2. Create the key dummy variable
combined['far_right'] = (combined['cps21_lr_scale_bef_1'] > 6).astype(int)

# 3. Basic controls (add more later if you want)
combined['age'] = combined['cps21_age']
combined['female'] = (combined['cps21_genderid'] == 2).astype(int)   # 2 = female in this dataset
combined['urban'] = (combined['pes21_rural_urban'] >= 4).astype(int) # adjust threshold as needed
combined['rel_important'] = combined['cps21_rel_imp']                # if you filtered it

# 4. Simple regression (minimal work)
model1 = smf.ols('pes21_mhealth ~ far_right', data=combined).fit()
print(model1.summary())

# 5. With basic controls (still very easy)
model2 = smf.ols('pes21_mhealth ~ far_right + age + female + urban + rel_important', 
                 data=combined).fit()
print(model2.summary())
```
